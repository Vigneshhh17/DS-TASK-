```python
#Freuent vs Infrequent
freq_threshold = daily.groupby("Account")["trade_count"].mean().median()
acc_freq = daily.groupby("Account")["trade_count"].mean().reset_index()
acc_freq.columns = ["Account", "avg_trades"]
acc_freq["frequent"] = acc_freq["avg_trades"] >= freq_threshold
daily = daily.merge(acc_freq[["Account", "frequent"]], on="Account")
print(daily["frequent"].value_counts())
```

```python
daily.groupby(["classification", "frequent"])["daily_pnl"].mean().unstack().plot(kind="bar")
plt.title("Frequent vs Infrequent Traders PnL by Sentiment")
plt.legend(["Infrequent", "Frequent"])
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
acc_consist = daily.groupby("Account")["win_rate"].std().reset_index()
acc_consist.columns = ["Account", "win_rate_std"]
acc_consist["consistent"] = acc_consist["win_rate_std"] <= acc_consist["win_rate_std"].median()
daily = daily.merge(acc_consist[["Account", "consistent"]], on="Account")
print(daily["consistent"].value_counts())
```

```python
daily.groupby(["classification", "consistent"])["daily_pnl"].mean().unstack().plot(kind="bar")
plt.title("Consistent vs Inconsistent Traders PnL by Sentiment")
plt.legend(["Inconsistent", "Consistent"])
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

**PART C** - **ACTIONABLE OUTPUT**


- On Greed Days — Non-Profitable Traders Should Reduce Exposure
- Non-profitable traders average -14k PnL on Greed days
- Profitable traders stay positive across all sentiments
- Rule: If sentiment = Greed/Extreme Greed, reduce position size by 50%
  and avoid opening new longs unless win rate > 60% in last 7 days
- Follow Consistent Trader Behavior Regardless of Sentiment
- Consistent traders maintain stable win rates even during Fear/Extreme Fear
- Inconsistent traders performance swings wildly with sentiment
- Rule: Cap daily trade count to your personal median,
  do not increase frequency during Extreme Fear or Extreme Greed
